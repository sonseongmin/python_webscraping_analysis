5. 함수 이해하기

```python
#1  두 수와 연산자(+, -, *, /)를 인자로 받아 계산하는 calculator 함수를 작성하세요.  
def cal(a,b,op):
   
    if op == '+':
        return a + b
    elif op == '-':
        return a - b
    elif op == '*':
        return a * b
    elif op == '/':
        return a / b
    else: return "Error"

#try:
    a=int(input('숫자입력'))
    op=input('연산 뭐 할예?')
    b=int(input('하나 더 입력'))
    result=cal(a,b,op)
    print(result)
#except ValueError:
    print('힝')

try:
    op=input('연산 뭐 할예?')
    nums=[1,2,3,4,5]
    res= [cal(x,x,op) for x in nums]
    print(f'결과:{res}')
except ValueError:
    print('힝')
    
```

```python
#2 문자열을 인자로 받아 단어 수를 반환하는 count_words 함수를 작성하세요.  
def count_words(sentence):
    print(f'입력한문장 {sentence}')
    return len(sentence.replace(" ",""))
text=input('함 입력해봐라')
print(count_words(text))
```

```python
#3 시작값, 끝값을 인자로 받아 그 사이의 소수를 리스트로 반환하는 find_primes 함수를 작성하세요.
import math

def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True

def find_primes(start, end):
    primes = []
    for num in range(start, end + 1): 
        if is_prime(num):
            primes.append(num) 
    return primes

start = int(input("시작값을 입력하세요: "))
end = int(input("끝값을 입력하세요: "))
print(find_primes(start, end))
```

6. Pythonic Code 이해하기

```python

#1 리스트 [1, 2, 3, 4, 5]의 각 요소를 제곱한 새 리스트를 생성하세요. (List Comprehension 사용)
nums=[1,2,3,4,5]
noms=[x**2 for x in nums]
print(noms)
```

```python
#2  [10, 20, 30, 40, 50]에서 30보다 큰 수만 필터링하세요. (filter 함수 사용)
nums= [10,20,30,40,50]
wow=filter(lambda x: x > 30, nums)
print(list(wow))
```

```python
#3 두 리스트 [1, 2, 3]과 [10, 20, 30]의 요소들을 짝지어 출력하세요. (zip 함수 사용)

ddong = [1,2,3]
ojoom = [10,20,30]
for a,b in zip(ddong, ojoom):
    print(a,b)
```

```python
#4 [1, 2, 3, 4, 5]의 각 요소에 10을 더한 후 2로 나눈 결과를 리스트로 생성하세요. (리스트 컴프리헨션 사용)
nums= [1,2,3,4,5]
res=[(x+10)/2 for x in nums]
print(res)
```

```python
#5 ["apple", "banana", "cherry"]에서 각 단어의 길이를 딕셔너리로 만드세요. (딕셔너리 컴프리헨션 사용)
fruit = ["apple", "banana", "cherry"]
length = {fruit:len(fruit) for fruit in fruit}
print(length)
```
